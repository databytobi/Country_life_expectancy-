```python
# core libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR

from sklearn.metrics import mean_absolute_error

from functions import *
```

perpare the data

```python
y = df["lifeexp"]
X = df[["happiness", "income", "sanitation", "water", "literacy", "inequality", "energy", "childmortality", "fertility", "hiv", "foodsupply", "population"]]
```

scale the features

```python
scaler = MinMaxScaler(feature_range=(0,1))
rescaledx = scaler.fit_transform(X)

X = pd.DataFrame(rescaledx, index=X.index,columns=X.columns)
```

build model

```python
test_size = 0.33
Seed = 1
X_train, X_test, Y_train, Y_test = train_test_split(X, y, test_size=test_size, random_state=Seed)
```

create muliple model 

```python
models = [LinearRegression(), KNeighborsRegressor(), SVR()]
```

```python
for model in models:
    model.fit(X_train,Y_train)
    predictions = model.predict(X_train)
    print(type(model).__name__, mean_absolute_error(Y_train, predictions))
```

Evalute the models

```python
for model in models:
    predictions = model.predict(X_test)
    print(type(model).__name__, mean_absolute_error(Y_test, predictions))
```

```python


def LinearRegressionSummary(model, column_names):

    coef = model.coef_.flatten() if model.coef_.ndim > 1 else model.coef_

    fig, ax = plt.subplots(figsize=(10, len(column_names) / 3))
    fig.suptitle("Linear Regression coefficients", fontsize= 16)

    ax.barh(column_names, coef, color="skyblue")

    for rect in ax.patches:
        width = rect.get_width()
        ax.annotate(f'{width:.4f}',
                     xy=(rect.get_x() + width, rect.get_y()),
                       xytext=(5, 0),
                         textcoords= "offset points",
                    ha = "left" if width > 0 else "right", va="bottom")
```

```python
LinearRegressionSummary(models[0], list(X_train))
```
